```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image (change path accordingly)
image = cv2.imread("camera.png")

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
else:
    print("Image loaded successfully")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
```

```python
height, width, channels = image.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
print("Data type:", image.dtype)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
```

```python
# Access pixel value at (row=100, col=100)
pixel_value = gray[100, 100]
print("Pixel value at (100,100):", pixel_value)

# Modify pixel value
gray[100, 100] = 255
```

```python
resized = cv2.resize(image_rgb, (300, 300))

plt.imshow(resized)
plt.title("Resized Image (300x300)")
plt.axis("off")
```

```python
# Increase brightness
bright = cv2.convertScaleAbs(image_rgb, alpha=1, beta=50)

# Increase contrast
contrast = cv2.convertScaleAbs(image_rgb, alpha=1.5, beta=0)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(bright)
plt.title("Brightened")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(contrast)
plt.title("High Contrast")
plt.axis("off")
```
